Całkowanie numeryczne jest szeroko stosowane w praktyce, szczególnie w obszarach takich jak nauki przyrodnicze, inżynieria, ekonometria, czy analiza danych.

Podstawowe metody całkowania numerycznego obejmują:

**[Metoda prostokątów](https://en.wikipedia.org/wiki/Riemann_sum#Left_rule)**

Metoda prostokątów jest najprostszą z metod kwadraturowych i polega na przybliżeniu całki poprzez sumowanie wartości funkcji w wybranych punktach podprzedziałów i pomnożeniu tej sumy przez szerokość podprzedziału.

$$
\int_{x_0}^{x_n}f(x)dx\approx h\sum_{i=0}^{n-1}f(x_i + \alpha h)
$$

gdzie:
$$
h = \frac{x_n-x_0}{n}
$$

Podczas laboratorium rozpatrywać będziemy wariant lewych prostokątów, gdy $\alpha = 0$.

**[Metoda trapezów](https://en.wikipedia.org/wiki/Trapezoidal_rule)**

Metoda trapezów działa analogicznie jak metoda prostokątów, ale używa trapezów łącząc tym samym wartości funkcji na końcach podprzedziałów. Dzięki temu daje ona lepsze rezultaty.

$$
\int_{x_0}^{x_n}f(x)dx\approx \sum_{i=0}^{n-1}\frac{h}{2}\big(f(x_{i+1})+f(x_i)\big)
$$
